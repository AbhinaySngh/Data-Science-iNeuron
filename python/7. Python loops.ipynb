# AND Condition
In this condition both the statement is true. 

```python
a = 10
b = 20
# In "and" both the statement is true.
# if a == 10 and b == 20:
if a == 10 and b > 20:
    print("this is a valid statement")
elif a == 10 and b < 20:
    print("this statement is valid inside elif")
else:
    print("condition is not valid")
```

```python
a == 10 and b == 20
```

# OR Condition
In this condition either one or both the statement is true.

```python
a = 10
b = 20

if a == 10 or b > 20:
    print("this is a valid statement")
elif a == 10 and b < 20:
    print("this statement is valid inside elif")
else:
    print("condition is not valid")
```

```python
a == 10 or b > 20
```

```python
def pattern(n):
    k = 2*n-2
    for i in range(n,-1,-1):
        for j in range(0,k,+1):
            print(end=" ")
        k = k+1
        for j in range(0,i+1):
            print("*",end=" ")
        print('\r')
            
pattern(5)
```

```python
n = int(input("Enter number of rows required : "))

for i in range(n,0,-1):
    for j in range(n-i):
        print(" ",end=" ")
        
    for j in range(2*i-1):
        print("*",end=" ")
    print() #print("\r") both is used here to go on next line.
```

```python
userInput = int(input("Please enter the amount of rows: "))

row = 0
while(row < userInput):
    row += 1
    spaces = userInput - row

    spaces_counter = 0
    while(spaces_counter < spaces):
        print(" ", end='')
        spaces_counter += 1

    num_stars = 2*row-1
    while(num_stars > 0):
        print("*", end='')
        num_stars -= 1

    print()
```
