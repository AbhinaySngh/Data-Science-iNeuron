```python
name = "Abhinay"
vowels = "AaEeIiOoUu"

for i in name :
    if i in vowels:
        print("{} is a vowel".format(i))
    else:
        print("{} is not a vowel".format(i))
```

```python
# Palindrome
s = input("enter a dat: ")
s1 = s[::-1]
if s == s1:
    print("It is palindrome")
else:
    print("It is not palindrome")
```
